### Create geneAnnotation file

```r
suppressPackageStartupMessages({
        library(biomaRt)
        library(BSgenome.Ocuniculus.NCBI.oryCun2)
        library(GenomicFeatures)
        library(compEpiTools)
        library(GenomeInfoDb)
        library(stringi)
        library(AnnotationHub)
    })
```

```r
# Create gene annotation using Txdb file
rabbit_txdb <- makeTxDbFromBiomart(biomart="ensembl", dataset="ocuniculus_gene_ensembl")

taxid = taxonomyId(rabbit_txdb)
## If we don't have a package, then lets get the taxIds and AHIds
        ## for the hub objects
        loadNamespace("AnnotationHub")
        ah <- AnnotationHub::AnnotationHub()
        ah <- subset(ah, ah$rdataclass=='OrgDb') 
        mc <- mcols(ah)[,'taxonomyid', drop=FALSE]
        ## Then just get the object
        AHID <- rownames(mc[mc$taxonomyid==taxid,,drop=FALSE])
        rabbit_OrgDb <- ah[[AHID[2]]]

#rabbit_txdb
geneAnnotation <- createGeneAnnotation(
  TxDb  = rabbit_txdb,
    OrgDb =rabbit_OrgDb,
  TSS = TSS(rabbit_txdb), 
  exons = exons(rabbit_txdb)
)
```

```r
chrnames = fread('chr_names.txt', sep='\t')[,1:2]
colnames(chrnames) = paste0('V', c(2,1))
```

```r
# seqnames order
chr_old = data.frame(V1 = as.vector(seqnames(geneAnnotation$genes)))
chr_old$idx = 1:nrow(chr_old)
chr_new = merge(chr_old, chrnames, by='V1')
chr_new = chr_new[order(chr_new$idx),]
```

```r
anno_gene.GR = GRanges(
    seqnames = Rle(chr_new$V2),
    ranges = IRanges(start= geneAnnotation$genes@ranges@start, 
                     end = geneAnnotation$genes@ranges@start + geneAnnotation$genes@ranges@width),
    strand = Rle(as.vector(geneAnnotation$genes@strand)),
    gene_id = geneAnnotation$genes$gene_id)

# gene names
order = as.data.frame(anno_gene.GR)
order$idx = 1:nrow(order)
symbols = merge(order, final_gene_names, by='gene_id', all.x=TRUE)
symbols = symbols[order(symbols$idx),]
symbols[is.na(symbols$symbol),]$symbol = symbols[is.na(symbols$symbol),]$gene_id

anno_gene.GR$gene_name = symbols$symbol
```

```r
# Annotate exons with genes
overlaps = as.data.frame(findOverlaps(geneAnnotation$exons, geneAnnotation$genes))
overlaps = overlaps[!duplicated(overlaps$queryHits),] # removes exons mapped to multiple genes
exon_names = geneAnnotation$genes$gene_id[overlaps$subjectHits]

# seqnames order
chr_old = data.frame(V1 = as.vector(seqnames(geneAnnotation$exons)))
chr_old$idx = 1:nrow(chr_old)
chr_new = merge(chr_old, chrnames, by='V1')
chr_new = chr_new[order(chr_new$idx),]
```

```r
anno_exon.GR = GRanges(
    seqnames = Rle(chr_new$V2),
    ranges = IRanges(start= geneAnnotation$exons@ranges@start, 
                     end = geneAnnotation$exons@ranges@start + geneAnnotation$exons@ranges@width),
    strand = Rle(as.vector(geneAnnotation$exons@strand)),
    exon_id = geneAnnotation$exons$exon_id, 
    gene_id = exon_names)

# gene names
order = as.data.frame(anno_exon.GR)
order$idx = 1:nrow(order)
symbols = merge(order, final_gene_names, by='gene_id', all.x=TRUE)
symbols = symbols[order(symbols$idx),]
symbols[is.na(symbols$symbol),]$symbol = symbols[is.na(symbols$symbol),]$gene_id

anno_exon.GR$gene_name = symbols$symbol
```

```r
# seqnames order
chr_old = data.frame(V1 = as.vector(seqnames(geneAnnotation$TSS)))
chr_old$idx = 1:nrow(chr_old)
chr_new = merge(chr_old, chrnames, by='V1')
chr_new = chr_new[order(chr_new$idx),]
```

```r
anno_TSS.GR = GRanges(
    seqnames = Rle(chr_new$V2),
    ranges = IRanges(start= geneAnnotation$TSS@ranges@start,
                    end = geneAnnotation$TSS@ranges@start, names = ),
    strand = Rle(as.vector(geneAnnotation$TSS@strand)),
    tx_id = geneAnnotation$TSS$tx_id, 
    tx_name =  geneAnnotation$TSS$tx_name)
```

```r
geneAnno = geneAnnotation
geneAnno$genes = anno_gene.GR
geneAnno$exons = anno_exon.GR
geneAnno$TSS = anno_TSS.GR
```

```r
saveRDS(geneAnno, "/rds/project/rds-SDzz0CATGms/users/bt392/04_Rabbit_ATAC_final/geneAnnotation_new.rds")
```

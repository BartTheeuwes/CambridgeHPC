# Create Peak Matrix

```r
source(here::here("load_archr.R"))
```

```r
ArchRProject@cellColData = ArchRProject@cellColData[,c(1:16, 23, 33:35)]
```

```r
# args
args = list()
args$group_by = 'Clusters_TileMatrix'
args$pathToMacs2 = findMacs2()
args$pvalue_cutoff = 0.001
args$extend_summits = 300
args$outdir = file.path(io$base, 'ArchR/Peak_calling')

dir.create(args$outdir, showWarnings = FALSE)
```

```r
addArchRThreads(threads = 15) 
```

```r
# Add Group Coverages
ArchRProject <- addGroupCoverages(ArchRProject, 
  groupBy = args$group_by,
  useLabels = TRUE,
  minCells = 150,
  maxCells = 1000,
  minReplicates = 2,
  maxReplicates = 7,
  force = TRUE
)

saveArchRProject(ArchRProject)
```

```r
# Call peaks
ArchRProject <- addReproduciblePeakSet(
  ArchRProj = ArchRProject, 
  groupBy = args$group_by, 
  peakMethod = "Macs2",
  excludeChr = c("chrM", "chrY"),
  pathToMacs2 = args$pathToMacs2,
  genomeSize = 2.7e9, # This argument is needed for the non-standard genome
  cutOff = args$pvalue_cutoff,
  extendSummits = args$extend_summits,
  plot = FALSE,
  force = TRUE
)

# Save
saveArchRProject(ArchRProject)
# fetch peaks in data.table format
dt <- getPeakSet(ArchRProject) %>% as.data.table() %>% setnames("seqnames","chr")
# Save peak metadata
fwrite(dt, file.path(args$outdir,"peak_metadata.tsv.gz"), sep="\t")
# save peaks in bed format
fwrite(dt[,c("chr","start","end")], file.path(args$outdir,"peaks_archR_macs2.bed.gz"), sep="\t", col.names = F)
```

```r
ggplot(as.data.table(getPeakSet(ArchRProject)), aes(peakType,fill=peakType)) + geom_bar() + theme_bw() + theme(text=element_text(size=20))
```

```r
plotBrowserTrack(
        ArchRProj = ArchRProject, 
        groupBy = "Clusters_TileMatrix", 
        geneSymbol = 'GATA3', 
        upstream = 50000,
        downstream = 50000,
        useMatrix = 'GeneIntegrationMatrix_TileMatrix', verbose=FALSE
    ) 
```

```r
# Add PeakMatrix
ArchRProject <- addPeakMatrix(ArchRProject)
# Save
saveArchRProject(ArchRProject)
```

# Compare mapping between Tile & Peak Matrices

```r
source(here::here("load_archr.R"))
```

```r
args = list()
```

```r
plot_embedding = function(matrix){
    args$matrix = matrix
    args$colour_by = c('Clusters_TileMatrix', 'Clusters_PeakMatrix', 
                       'predictedGroup_celltype_TileMatrix', 'predictedGroup_celltype_PeakMatrix')
    # Fetch UMAP coordinates
    umap.dt <- getEmbedding(ArchRProject,paste0("UMAP_", args$matrix)) %>%
      round(2) %>%
      as.data.table(keep.rownames = T) %>%
      setnames(c("cell","umap1","umap2"))

        # Plot
    sample_metadata = as.data.frame(getCellColData(ArchRProject))
    to.plot <- umap.dt %>%
      merge(sample_metadata,by="cell")
    

    p = list()
    
    for (k in args$colour_by) {
        p[[k]] <- ggplot(to.plot, aes_string(x="umap1", y="umap2", fill=k)) +
            geom_point(size=1.5, shape=21, stroke=0.05) + 
             scale_fill_manual(values = as.vector(c(ArchRPalettes[[3]],ArchRPalettes[[5]], ArchRPalettes[[8]],ArchRPalettes[[9]]))[1:length(unique(to.plot[[k]]))]) + 
            theme_classic() + theme(legend.position='none')  
    }
    ggarrange(p[[args$colour_by[1]]],p[[args$colour_by[2]]],p[[args$colour_by[3]]],p[[args$colour_by[4]]], ncol=2)
}
```

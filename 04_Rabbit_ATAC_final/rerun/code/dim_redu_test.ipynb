# Try with ArchR Peakset

```r
dim_redu = function(dims, features){
# Iterative LSI: two iterations
ArchRProject <- addIterativeLSI(
  ArchRProj = ArchRProject,
  useMatrix = 'PeakMatrix', 
  name = 'LSI', 
  dimsToUse = 1:dims,
  firstSelection = "Top",
  depthCol = "nFrags",
  iterations = 2, 
  saveIterations = FALSE,
  varFeatures = features, 
  force = TRUE, verbose=FALSE
)

# Run UMAP
ArchRProject <- addUMAP(
  ArchRProj = ArchRProject, 
  reducedDims = 'LSI',
  name = "UMAP_PeakMatrix",
  metric = "cosine",
  nNeighbors = 25, 
  minDist = 0.3, 
  seed = 45,
  saveModel = FALSE,
  force = TRUE, verbose=FALSE
)

p = plotEmbedding(ArchRProject, 
'UMAP_PeakMatrix', 
colorBy = "cellColData", 
name = 'celltype_manual')
    print(p)
}
```

```r
in_path = '/rds/project/rds-SDzz0CATGms/users/bt392/04_Rabbit_ATAC_final/data/raw/Combined'
```

```r
Peaks = GRanges(
    seqnames = Rle(),
    ranges = IRanges(), origin='')
```

```r
samples = 1:8
Peaks = GRanges(
    seqnames = Rle(),
    ranges = IRanges(), 
    origin='')

for(i in samples){
    print(paste('sample =', i))
    # read peaks
    add_peaks = fread(sprintf('%s/BGRGP%s/outs/peak_annotation.tsv', in_path, i))
    # Create Granges object
    add_peaks.gr = GRanges(
        seqnames = Rle(paste0('chr',add_peaks$chrom)),
        ranges = IRanges(add_peaks$start, end = add_peaks$end, names = add_peaks$gene),
        origin = i)
    # Find overlap with others
    overlap = findOverlaps(Peaks, add_peaks.gr,  minoverlap = 200)
    print(paste('overlap:', length(overlap)))
    # Remove regions already present
    if(length(overlap)>0){
        add_peaks.gr = add_peaks.gr[-subjectHits(overlap),]
    }
    print(paste('new:', length(add_peaks.gr)))
    # Add old + new peaks
    Peaks = do.call('c', list(Peaks, add_peaks.gr))
}
```

```r
# Vast majority of peaks comes from the first sample (GD7)
```

```r
ArchRProject = addPeakSet(ArchRProject, Peaks, force=TRUE)
```

```r
ArchRProject <- addPeakMatrix(ArchRProject, force=TRUE)
```

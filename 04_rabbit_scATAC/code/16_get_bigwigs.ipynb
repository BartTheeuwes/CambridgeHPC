```r
opts$bigwig_colours = opts$celltype_colours[names(opts$celltype_colours) %in% unique(ArchRProject$predictedGroup_celltype_PeakMatrix)]
opts$bigwig_colours = opts$bigwig_colours[order(names(opts$bigwig_colours))]
opts$bigwig_colours = opts$bigwig_colours[!duplicated(opts$bigwig_colours)]
names(opts$bigwig_colours) = gsub('/', '.', gsub(' ', '.', gsub('-', '.', names(opts$bigwig_colours))))
```

```r
files = list.files(sprintf('%s/GroupBigWigs/predictedGroup_celltype_PeakMatrix/',io$archR.directory), pattern='.bw')
```

```r
names(files) = stringr::str_split(files, '-') %>% map_chr(1)
```

```r
opts$bigwig_colours = opts$bigwig_colours[names(opts$bigwig_colours) %in% names(files)]
```

```r
ucsc = lapply(1:length(opts$bigwig_colours), function(x){ # length(opts$bigwig_colours)
    ct = names(opts$bigwig_colours)[x]
    paste0('track type=bigWig name="', ct, 
           '" description="', ct, 
           '" color=', paste(as.vector(col2rgb(opts$bigwig_colours[x])), collapse = ","), 
           ' visibility=0 bigDataUrl=https://bioinformatics.stemcells.cam.ac.uk/Files_for_transfer/Open/bart/04_rabbit_scATAC/bigwigs/predictedGroup_celltype_PeakMatrix/', files[[ct]])
}) #%>% unlist()
```

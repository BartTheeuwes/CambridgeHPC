```r
library(data.table)
library(tidyr)
```

```r
files = list.files(path = ".", pattern = ".gz")
samples = unique(sapply(strsplit(files, '_'), '[', 1))
```

```r
head(files)
head(samples)
```

```r
file_meta = lapply(samples, function(x){
    tmp = files[grep(x, files)]
    print(paste0(x, ': ', length(tmp), ' files'))
    tmp = as.data.frame(tmp)
    colnames(tmp) = x
    return(as.data.frame(t(tmp)))
})
```

```r
file_meta = as.data.frame(rbindlist(file_meta))
rownames(file_meta) = samples
```

```r
file_meta
```

```r
fwrite(file_meta, 'file_meta.csv', quote=FALSE,  row.names = T, col.names = F)
```

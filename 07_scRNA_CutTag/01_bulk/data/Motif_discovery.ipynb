```python
TF = 'Cebpa'

TF.gr = fread(paste0(getwd(), '/X204SC22051679-Z01-F001/raw_data/', TF, '/macs2/NA_summits.bed')) %>%
    .[,c('V1', 'V2', 'V3')] %>% 
    setnames(c('chr', 'start', 'end')) %>%
    makeGRangesFromDataFrame %>% resize(., width = 100, fix='center')

TF.seq = getSeq(Mmusculus, TF.gr)
```

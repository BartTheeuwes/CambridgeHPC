```r
library(reticulate)
np <- import("numpy")
```

```r
contributions <- np$load("/rds/project/rds-SDzz0CATGms/users/bt392/09_Eomes_invitro_blood/results/atac/crested/v2/trial1/contributions/Allantois_contrib.npz")
contributions = array(contributions$f["arr_0"], dim = c(10000, 4, 2114)) # first one is number of peaks, thirds is length of prediction

sequences <- np$load("/rds/project/rds-SDzz0CATGms/users/bt392/09_Eomes_invitro_blood/results/atac/crested/v2/trial1/contributions/Allantois_oh.npz")
sequences = array(sequences$f["arr_0"], dim = c(10000, 4, 2114))

allantois = contributions * sequences
```

```r
plot_peak = function(peak_nr, width = 250){
    tmp1 = allantois[peak_nr,,]
    rownames(tmp1) = c('A', 'C', 'G', 'T')
    
    start_plot = 2114/2 - width
    end_plot = 2114/2 + width
    
    ggseqlogo(tmp1, method='custom', seq_type='dna') + ylab('my custom height') + 
            geom_hline(yintercept = 0) +
                theme_bw() + 
                ylim(min(c(tmp1)), max(c(tmp1))) + 
                scale_x_continuous(expand = c(0,0), limits = c(start_plot,end_plot)) + 
                theme(panel.grid = element_blank(), 
                      panel.background = element_rect(color = 'black', fill = NA, linewidth = 1),
                      axis.text.y = element_text(size = 15),
                     axis.title.y = element_blank(),
                     axis.text.x = element_blank(),
                     axis.ticks.x = element_blank(),
                     legend.position = 'none')
}

options(repr.plot.width = 20, repr.plot.height = 5)
mclapply(100:200, plot_peak, mc.cores = 42)
```

# Export supplementary table of DEGs and DARs

```r
here::i_am("rna_atac/differential/v2/01_early_defects.ipynb")

source(here::here("settings.R"))
source(here::here("utils.R"))
```

```r
args = list()
# Load Differential analysis
args$diffdir = file.path(io$basedir, 'results/rna_atac/differential/v2/')
args$expression = file.path(io$basedir, 'results/rna_atac/differential/v2/expression/')
args$accessibility = file.path(io$basedir, 'results/rna_atac/differential/v2/Accessibility/')
```

```r
opts$celltype_v2_figure
```

```r
# Load differential RNA
DEGs = fread(sprintf('%s/DEGs_edgeR_pb.txt.gz', args$expression)) %>% 
        .[,sig:=ifelse(abs(logFC)>0.5 & padj_fdr < 0.05, TRUE, FALSE)] %>% 
        .[, celltype := opts$celltype_v2_figure[match(celltype, names(opts$celltype_v2_figure))]] %>%
    setnames(c('mean_groupA', 'mean_groupB', 'cdr_groupA', 'cdr_groupB'),
             c('mean_WT', 'mean_EomesKD', 'cdr_WT', 'cdr_EomesKD')) %>%
    .[,c('celltype', 'gene', 'logFC', 'padj_fdr', 'mean_WT', 'mean_EomesKD', 'cdr_WT', 'cdr_EomesKD', 'sig')]

# Load differential ATAC
DARs = fread(sprintf('%s/DARs_edgeR_pb.txt.gz', args$accessibility)) %>% 
        .[,sig:=ifelse(abs(logFC)>0.5 & padj_fdr < 0.05, TRUE, FALSE)] %>% 
        .[, celltype := opts$celltype_v2_figure[match(celltype, names(opts$celltype_v2_figure))]] %>%
    setnames(c('mean_groupA', 'mean_groupB', 'cdr_groupA', 'cdr_groupB'),
             c('mean_WT', 'mean_EomesKD', 'cdr_WT', 'cdr_EomesKD')) %>%
    .[,c('celltype', 'peak', 'logFC', 'padj_fdr', 'mean_WT', 'mean_EomesKD', 'cdr_WT', 'cdr_EomesKD', 'sig')]
```

```r
fwrite(DEGs, sprintf('%s/Stable_invitro_DEGs.csv', args$diffdir), sep = ',')
fwrite(DARs, sprintf('%s/Stable_invitro_DARs.csv', args$diffdir), sep = ',')
```

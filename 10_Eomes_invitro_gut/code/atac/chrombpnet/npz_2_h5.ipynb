```python
import numpy as np
import h5py

# Load the .npz file
npz_file = np.load("/rds/project/rds-SDzz0CATGms/users/bt392/10_Eomes_invitro_gut/results/atac/chrombpnet/PS_EOi/predictions/contributions.npz")

# List all available files inside the npz archive
print("Files in npz:", npz_file.files)


# Extract a specific array (replace 'your_array' with the correct key)
array_name = "arr_0"  # Change this to the name of the array you want
data = npz_file[array_name]

# Save as an HDF5 (.h5) file
with h5py.File("/rds/project/rds-SDzz0CATGms/users/bt392/10_Eomes_invitro_gut/results/atac/chrombpnet/PS_EOi/predictions/contributions.h5", "w") as h5_file:
    h5_file.create_dataset(array_name, data=data)

print(f"Saved {array_name} to output.h5")
```

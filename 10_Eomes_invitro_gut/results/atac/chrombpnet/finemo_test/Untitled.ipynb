```python
import numpy as np

# Load the .npz file
data = np.load("shap.npz")

# List all keys (variable names stored inside)
print("Keys in .npz file:", data.files)
```

```python
# Iterate through each key and print the first few lines
for key in data.files:
    print(f"Key: {key}")
    print(data[key][:5])  # Show first 5 rows (adjust as needed)
    print("-" * 40)  # Separator for clarity
```

```python
# Save each key as a separate .npz file
for key in data.files:
    np.savez(f"{key}.npz", arr_0=data[key])  # Rename stored array to "arr_0"
    print(f"Saved {key}.npz with key renamed to 'arr_0'")
```

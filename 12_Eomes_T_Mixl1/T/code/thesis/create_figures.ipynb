```r
args = list()
args$metadata = paste0(io$basedir, 'results/rna/mapping/manual/metadata_mapping_updated.txt.gz')
args$outdir = paste0(io$basedir,"/results/thesis/")
dir.create(args$outdir, recursive=TRUE, showWarnings = FALSE)
```

```r
meta = fread(args$metadata)
atlas_meta = fread(io$rna.atlas.metadata)
```

```r
library(ggrastr)
```

```r
p1 = ggplot(meta[tdTom == F], aes(umapX, umapY)) + 
    ggrastr::rasterise(geom_point(data = atlas_meta, color = 'grey90', size = 0.2), dpi = 600) +
    ggrastr::rasterise(geom_point(size = -0.2, color = 'black'), dpi = 600) +
    ggtitle('WT') +
    coord_fixed(ratio = 0.5) +
    theme_void() +
    theme(plot.title = element_text(size = 20, hjust = 0.5))

p2 = ggplot(meta[tdTom == T], aes(umapX, umapY)) + 
    ggrastr::rasterise(geom_point(data = atlas_meta, color = 'grey90', size = 0.2), dpi = 600) +
    ggrastr::rasterise(geom_point(size = -0.2, color = 'red'), dpi = 600) +
    ggtitle('T KO') +
    coord_fixed(ratio = 0.5) +
    theme_void() +
    theme(plot.title = element_text(size = 20, hjust = 0.5))

# ggarrange(p1, p2, align = 'hv')

ggsave(file.path(args$outdir, 'UMAPmapping.pdf'),
       plot = ggarrange(p1, p2, align = 'hv'),
       width = 130, 
       height = 85, 
       units = "mm")
```

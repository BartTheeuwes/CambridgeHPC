```r
library(data.table)
library(stringr)
```

```r
# Load data from https://raw.githubusercontent.com/GundryLab/SurfaceGenie/master/ref/annotation.mouse.tsv
markers = fread("SurfaceGenie_SPC.txt")

# Only keep genes with SPC score > 0
markers = markers[SPC>0]

# list all genes (also keeping alternative names in case these are the ones used in mm10)
markers_names = unique(unlist(str_split(markers$geneName, " ")))

# Save list
write.table(markers_names, 'SurfaceGenie_SPC_geneNames.txt', quote = F, row.names = F, col.names = F)
```

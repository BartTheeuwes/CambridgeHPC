# Prepare extended atlas for blood cell types

```r
suppressPackageStartupMessages({
    library(data.table)
    library(Matrix)
    library(dplyr)
    library(SingleCellExperiment)
})
```

```r
main = '/rds/project/rds-SDzz0CATGms/users/bt392/atlasses/extended/'
```

```r
sce = readRDS(file.path(main, 'embryo_sce.rds'))
```

```r
sce
```

```r
ct_keep = c('Haematoendothelial progenitors',
  'Blood progenitors',
  'Allantois endothelium', 
  'Embryo proper endothelium',
  'Venous endothelium',
  'EMP',
  'YS endothelium', 
  'Erythroid',
  'MEP',
  'Megakaryocyte progenitors',
  'Chorioallantoic-derived erythroid progenitors',
  'YS mesothelium-derived endothelial progenitors')
```

```r
# Get metadata
meta = as.data.table(colData(sce))[celltype_extended_atlas %in% ct_keep]
```

```r
# Save
fwrite(meta, file.path(main, 'blood/meta_blood.txt.gz'))
```

```r
# number of cells
nrow(meta)
```

```r
sce_blood = sce[,as.character(meta$cell)]
```

```r
saveRDS(sce_blood, file.path(main, 'blood/blood_sce.rds'))
```

```python
import scanpy as sc
import pandas as pd
import numpy as np
```

```python
e65 = sc.read('embryo_6.5_6.75.h5ad')
e675 = sc.read('embryo_6.75_7.0.h5ad')
e70 = sc.read('embryo_7.0_7.25.h5ad')
e725 = sc.read('embryo_7.25_7.5.h5ad')
e75 = sc.read('embryo_7.5_7.75.h5ad')
e775 = sc.read('embryo_7.75_8.0.h5ad')
e80 = sc.read('embryo_8.0_8.25.h5ad')
e825 = sc.read('embryo_8.25_8.5.h5ad')
```

```python
e65_df = pd.DataFrame(e65.X)
e675_df = pd.DataFrame(e675.X)
e70_df = pd.DataFrame(e70.X)
e725_df = pd.DataFrame(e725.X)
e75_df  = pd.DataFrame(e75.X)
e775_df = pd.DataFrame(e775.X)
e80_df = pd.DataFrame(e80.X)
e825_df = pd.DataFrame(e825.X)
```

```python
e65_df.index = e65.obs.index
e65_df.columns = e675.obs.index

e675_df.index = e675.obs.index
e675_df.columns = e70.obs.index

e70_df.index = e70.obs.index
e70_df.columns = e725.obs.index

e725_df.index = e725.obs.index
e725_df.columns = e75.obs.index

e75_df.index = e75.obs.index
e75_df.columns = e775.obs.index

e775_df.index = e775.obs.index
e775_df.columns = e80.obs.index

e80_df.index = e80.obs.index
e80_df.columns = e825.obs.index

# No column data available pfff
# e825_df.index = e65.obs.index
# e825_df.columns = e675.obs.index
```

```python
e65_cells = e65_df.apply(lambda x:  pd.Series(x.nlargest(5).index.tolist()), axis=1)
e675_cells = e675_df.apply(lambda x:  pd.Series(x.nlargest(5).index.tolist()), axis=1)
e70_cells = e70_df.apply(lambda x:  pd.Series(x.nlargest(5).index.tolist()), axis=1)
e725_cells = e725_df.apply(lambda x:  pd.Series(x.nlargest(5).index.tolist()), axis=1)
e75_cells = e75_df.apply(lambda x:  pd.Series(x.nlargest(5).index.tolist()), axis=1)
e775_cells = e775_df.apply(lambda x:  pd.Series(x.nlargest(5).index.tolist()), axis=1)
e80_cells = e80_df.apply(lambda x:  pd.Series(x.nlargest(5).index.tolist()), axis=1)
```

```python
all_transitions = result = pd.concat([e65_cells, 
                                      e675_cells, 
                                      e70_cells,
                                      e725_cells,
                                      e75_cells,
                                      e775_cells,
                                      e80_cells], axis=0)
```

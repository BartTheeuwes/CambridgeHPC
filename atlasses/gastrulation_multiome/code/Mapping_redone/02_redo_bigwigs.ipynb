```r
args = list() 
# ATAC
args$archr_directory <- file.path(io$basedir,"data/processed/atac/archR")
args$metadata <- file.path(io$basedir,"/results/atac/archR/qc/sample_metadata_after_qc.txt.gz")
```

```r
sample_metadata <- fread(args$metadata) %>%
  .[pass_atacQC==TRUE] 
```

```r
meta_new = fread('meta_new_mapping.txt.gz')
cells_keep = intersect(sample_metadata$cell, meta_new$cell)
meta_new = meta_new[match(cells_keep, cell)]
```

```r
table(meta_new$celltype.mapped)
```

```r
########################
## Load ArchR Project ##
########################

setwd(args$archr_directory)

addArchRGenome("mm10")
addArchRThreads(threads = detectCores()-2) 

ArchRProject <- loadArchRProject(args$archr_directory)

# Subset to cells that pass QC
ArchRProject <- ArchRProject[cells_keep]
```

```r
ArchRProject <- addCellColData(ArchRProj = ArchRProject, 
                                    data = as.character(meta_new$celltype.mapped),
                                    cells = cells_keep,
                                    name = "celltype_mapped_redone",
                                    force=TRUE)
```

```r
addArchRThreads(24)
getGroupBW(
  ArchRProj = ArchRProject,
  groupBy = 'celltype_mapped_redone',
  normMethod = c("ReadsInTSS"),
  tileSize = 100,
  maxCells = 1000, # default
  ceiling = 4
)
```

```python
chim_meta_in = main + 'chim_meta.tsv'
atlas_meta_in = main + 'atlas_meta.tsv'
```

```python
chim_meta = pd.read_csv(chim_meta_in, sep='\t')
atlas_meta = pd.read_csv(atlas_meta_in, sep='\t')
```

```python
chim_meta = chim_meta.set_index('cell')
atlas_meta = atlas_meta.set_index('cell')
####
####
### FIRST FILTER ONLY META CELLS THAT ARE ALSO IN ADATA OBJECT!
```

```python
chim_adata.obs = chim_meta
atlas_adata.obs = atlas_meta
```

```python
chim_adata.write(main + 'chim_adata.h5ad')
atlas_adata.write(main + 'atlas_adata.h5ad')
```

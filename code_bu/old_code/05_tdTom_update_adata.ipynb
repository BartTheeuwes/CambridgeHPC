### update metadatas

```python
sample_tdtom = pd.DataFrame({'sample': ['mixl_1','mixl_2','mixl_3','mixl_4','mixl_5','mixl_6'], 
                             'tdTom': ['neg', 'pos', 'neg', 'pos', 'neg', 'pos'] })
```

```python
# Data
adata = sc.read(Data)

adata.obs['cell'] = adata.obs.index
adata.obs = adata.obs.drop(columns='tdTom').merge(sample_tdtom, on='sample', how='left')
adata.obs.index = adata.obs['cell']
adata.obs = adata.obs.drop(columns='cell')

adata.write(Data)
```

```python
# Results
adata = sc.read(Results)

adata.obs['cell'] = adata.obs.index
adata.obs = adata.obs.drop(columns='tdTom').merge(sample_tdtom, on='sample', how='left')
adata.obs.index = adata.obs['cell']
adata.obs = adata.obs.drop(columns='cell')

adata.write(Results)
```

```python
# velocity results
adata = sc.read(velocity)

adata.obs['cell'] = adata.obs.index
adata.obs = adata.obs.drop(columns='tdTom').merge(sample_tdtom, on='sample', how='left')
adata.obs.index = adata.obs['cell']
adata.obs = adata.obs.drop(columns='cell')

adata.write(velocity)
```
